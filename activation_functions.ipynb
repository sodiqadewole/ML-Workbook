### ReLU Activation

```python
import torch

def relu(x):
    return torch.max(torch.zeros_like(x), x)

z = torch.randn(10)
print(f"Output before activation: {z}")

custom_relu = relu(z)
print()
print(f"Custom implementation: {custom_relu}")

# Compare with PyTorch implementation
original_relu = torch.nn.ReLU()
original = original_relu(z)

print(f"Pytorch ReLU: {original}")
```

### Leaky ReLU Activation

```python
def leakyReLU(x, negative_slope=0.01):
    return torch.max(torch.zeros_like(x), x) + negative_slope * torch.min(torch.zeros_like(x), x)


z = torch.randn(10)
print(f"Output before activation: {z}")

custom_leaky_relu = leakyReLU(z)
print()
print(f"Custom implementation: {custom_leaky_relu}")

# Compare with PyTorch implementation
original_leaky_relu = torch.nn.LeakyReLU(0.01)
original = original_leaky_relu(z)

print(f"Pytorch ReLU: {original}")

print(torch.equal(custom_leaky_relu, original))
```

### Sigmoid Activation

```python
def custom_sigmoid(x):
    return 1 / (1 + torch.exp(-x))

z = torch.randn(10)
print(f"Output before activation: {z}")

custom_sigmoid = custom_sigmoid(z)
print()
print(f"Custom implementation: {custom_sigmoid}")

# Compare with PyTorch implementation
original_sigmoid = torch.nn.Sigmoid()
original = original_sigmoid(z)

print(f"Pytorch ReLU: {original}")

print(torch.equal(custom_sigmoid, original))
```

### Softmax Activation

```python
def softmax_activation(x):
    return torch.exp(x) / torch.sum(torch.exp(x), dim=0)

z = torch.randn(12)
print(f"Z before activation: {z}")

# Apply custom softmax
custom_softmax = softmax_activation(z)
print(f"Custom Softmax: {custom_softmax}")

# Compare with Pytorch implemented softmax
original_softmax = torch.nn.Softmax(dim=-1)
original = original_softmax(z)
print(f"Original Softmax: {original}")
```

```python
# To Add stability to prevent numeric
def custom_softmax(x, dim=-1):
    """
    Applies softmax to an n-dimensional input tensor.
    Formula: exp(x_i) / sum(exp(x_i))
    """
    # 1. Find the maximum value along the target dimension for stability
    max_x = torch.max(x, dim=dim, keepdim=True).values

    # 2. Exponentiate the stabilized inputs
    exp_x = torch.exp(x-max_x)

    # 3. Normalize by the sum of exponents along that same dimension
    sum_exp_x = torch.sum(exp_x, dim=dim, keepdim=True)
    
    return exp_x / sum_exp_x

print(custom_softmax(z))
```

### Tanh Activation

```python
def custom_tanh(x):
    return (torch.exp(x) - torch.exp(-x)) / (torch.exp(x) + torch.exp(-x))

z1 = torch.randint(low=0, high=100, size=(10,))
print(f"hidden state before activation: {z}")

c_tanh = custom_tanh(z1)
print(f"Custom tanh: {c_tanh}")

# compare with PyTorch implementation
original_tanh = torch.nn.Tanh()
original = original_tanh(z1)
print(f"Pytorch implementation: {original}")
```

```python
### To mitigate Overflow
def custom_tanh(x):
    '''
    Formula: (exp(x) - exp(-x)) / (exp(x) + exp(-x))
    
    Stable Formula:
        For x >= 0: (1 - exp(-2x)) / (1 + exp(-2x))
        For x < 0: (exp(2x) - 1) / (exp(2x) + 1)
    '''
    # create mask for stable calculation bounds
    is_positive = (x >= 0).to(x.dtype)
    is_negative = 1.0 - is_positive

    # Calculate for positive inputs:
    exp_neg_2x = torch.exp(-2.0 * torch.abs(x))
    pos_res = (1.0 - exp_neg_2x) / (1.0 + exp_neg_2x)

    # Calculate for negative inputs: (exp(2x) - 1)
    exp_pos_2x = torch.exp(-2.0 * torch.abs(x)) # mathematically identical via reflecting symmetry
    neg_res = (exp_pos_2x - 1.0) / (exp_pos_2x + 1.0)

    return is_positive * pos_res + is_negative * neg_res

z1 = torch.randint(low=0, high=100, size=(10,))
print(f"hidden state before activation: {z}")

c_tanh = custom_tanh(z1)
print(f"Custom tanh: {c_tanh}")

# compare with PyTorch implementation
original_tanh = torch.nn.Tanh()
original = original_tanh(z1)
print(f"Pytorch implementation: {original}")
```
